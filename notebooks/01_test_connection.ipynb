```python
# Test 1: Check serverless compute is attached
print("Spark version:", spark.version)

# Test 2: Write a small test file to bronze container
storage_account = "stretailcdcproj"
bronze_path = f"abfss://bronze@{storage_account}.dfs.core.windows.net/test/"

test_data = [("hello", 1), ("world", 2)]
df_test = spark.createDataFrame(test_data, ["word", "value"])

df_test.write.mode("overwrite").format("delta").save(bronze_path)
print("✅ Write successful!")

# Test 3: Read it back
df_read = spark.read.format("delta").load(bronze_path)
df_read.show()
```
